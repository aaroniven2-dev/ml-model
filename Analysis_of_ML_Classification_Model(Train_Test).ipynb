```python
# Install required libraries
!pip install pandas openpyxl scikit-learn

# Import libraries
import pandas as pd
import numpy as np
import pickle

# Google Drive
from google.colab import drive, files

# Machine Learning models
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC

# Train-test evaluation
from sklearn.metrics import (
    confusion_matrix,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

print("All libraries imported successfully!")
```

```python
# Connect Google Drive to Google Colab
drive.mount('/content/drive')
```

```python
train_file = '/content/drive/MyDrive/TRAIN_AI4I2020_PDM_DATASET.xlsx'
test_file = '/content/drive/MyDrive/TEST_AI4I2020_PDM_DATASET.xlsx'
```

```python
# Read Training Excel file
train_df = pd.read_excel(train_file)

# Read Testing Excel file
test_df = pd.read_excel(test_file)

print("Training DataFrame created successfully!")
print("Testing DataFrame created successfully!")

print("\nTraining Dataset:")
display(train_df.head())

print("\nTesting Dataset:")
display(test_df.head())
```

```python
X_train = train_df.iloc[:, 0:5]

y_train = train_df.iloc[:, 5]

X_test = test_df.iloc[:, 0:5]

y_test = test_df.iloc[:, 5]

print("X_train shape:", X_train.shape)
print("y_train shape:", y_train.shape)

print("X_test shape:", X_test.shape)
print("y_test shape:", y_test.shape)

print("\nTraining Features (X_train):")
display(X_train.head())

print("\nTraining Labels (y_train):")
display(y_train.head())

print("\nTesting Features (X_test):")
display(X_test.head())

print("\nTesting Labels (y_test):")
display(y_test.head())
```

```python
# Create Random Forest model
random_forest_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

# Train the model using training data
random_forest_model.fit(X_train, y_train)

print("Random Forest model trained successfully!")
```

```python
# Create Decision Tree model
decision_tree_model = DecisionTreeClassifier(
    random_state=42
)

# Train the model
decision_tree_model.fit(X_train, y_train)

print("Decision Tree model trained successfully!")
```

```python
# Create SVM model
svm_model = SVC(
    kernel='rbf',
    random_state=42
)

# Train the model
svm_model.fit(X_train, y_train)

print("SVM model trained successfully!")
```

```python
y_pred_rf = random_forest_model.predict(X_test)

print("Random Forest predictions generated successfully!")
print(y_pred_rf[:10])
```

```python
y_pred_dt = decision_tree_model.predict(X_test)

print("Decision Tree predictions generated successfully!")
print(y_pred_dt[:10])
```

```python
y_pred_svm = svm_model.predict(X_test)

print("SVM predictions generated successfully!")
print(y_pred_svm[:10])
```

```python
def evaluate_model(model_name, y_test, y_pred):

    print("\n==========================================")
    print(model_name)
    print("==========================================")

    # Confusion Matrix
    cm = confusion_matrix(y_test, y_pred)

    print("\nConfusion Matrix:")
    print(cm)

    # Accuracy
    accuracy = accuracy_score(y_test, y_pred)

    # Precision
    precision = precision_score(
        y_test,
        y_pred,
        zero_division=0
    )

    # Recall
    recall = recall_score(
        y_test,
        y_pred,
        zero_division=0
    )

    # F1 Score
    f1 = f1_score(
        y_test,
        y_pred,
        zero_division=0
    )

    print("\nAccuracy :", accuracy)
    print("Precision:", precision)
    print("Recall   :", recall)
    print("F1-Score :", f1)

    return accuracy, precision, recall, f1
```

```python
rf_results = evaluate_model(
    "Random Forest",
    y_test,
    y_pred_rf
)
```

```python
dt_results = evaluate_model(
    "Decision Tree",
    y_test,
    y_pred_dt
)
```

```python
svm_results = evaluate_model(
    "SVM",
    y_test,
    y_pred_svm
)
```
